# Name(identifier) in python

- Name(identifier) 객체에 부여된 이름
    - 파이썬의 모든 것은 객체
    - 기본 개체에 접근하기 위한 방법

- Namespace
    - 변수의 적용 범위를 구분하여 사용
    -  Function(local namespace) < module(global namespace)(지역변수) < built-in namespace (전역변수)



```python
def outer_function():
    # b : local
    b = 20
    def inner_func():
    # c : nested local
        c = 30
        print(a)
    inner_func()

# a : global
a = 10
outer_function()
```

```python
# a의 들여쓰기가 각각 다르게 들어있음
# 코드 블럭이 다르다
# 전역변수랑 지역변수는 들여쓰기로 구분됌

def outer_function():
    a = 20  # 블럭안에서 새롭게 만들어진 a(변수) , 아래의 들여쓰기없는 a와는 다른 변수임.

    def inner_function():
        a = 30 # 지역변수(전역변수의 위치로 가져와서 사용할 수 없음)
        print('a =', a)

    inner_function()
    print('a =', a)   # 이름을 적어줘야 동작함


a = 10  # 전역 변수(들여쓰기가 안되어있는 변수)
outer_function()
print('a =', a)
```

```python
# 전역 변수를 사용하는 방법 ,"global"

def outer_function():
    global a  # global는 전역변수를 사용할 수 있게 해주는 키워드 , 변수 이름은 동일하게 적용해야함
    a = 20

    def inner_function():
        global a
        a = 30  # 들여쓰기가 많이 되어있는 변수가 먼저 사용됌
        print('a =', a)

    inner_function()
    print('a =', a)


a = 10  # 전역변수 (global로 선언시 가져오게 되는 변수)
outer_function()
print('a =', a)
```

```python
def outer_function():
    global b
    b = 20

a = 10
print(a)
outer_function() # Call the function to set global b
print(b)
```

# while loop, while loop + else

- 키워드를 주고 조건이 참이면 반복

 # 사용자가 숫자를 입력하지 않을때까지 입력하세요(무한반복)

```python
# while 반복 // 무한루프에 빠질 수 있기 때문에 사용 최소화

n = int(input("Enter n: "))

# initialize sum and counter
sum = 0
i = 1

while i <= n:
    sum = sum + i
    i = i+1 # update counter  #중요 종료할 수 있는 코드가 적어야함

# print the sum
print("The sum is", sum)
```

```python
counter = 0

while counter < 3:
    print("Inside loop")
    counter = counter + 1
else:
    print("Inside else")
```

```python
# filter(), map()동일한 연산 적용 , 등의 내장 함수에 사용

my_list = [1, 5, 4, 6, 8, 11, 3, 12]    # 숫자를 나열
new_list = list(filter(lambda x: (x%2 == 0) , my_list)) # lambda- 조건, my_list-입력
# filter 무언가를 걸러내기위한 것 ,  lambda라는 키워드를 주고 2로 나눴을때 0이 나오는 값

print(new_list)
```
