# Docstrings(documentation striongs)in python


- 함수에 대해 알고 싶을 때 ( 검색방법 )
- 코드의 동작을 문서화하는 데 쓰인다.

  - print (print.__doc__)
  - 이름 . __doc__
  - 파이썬자체에서 쉽게 검색하고 설명을 볼 수 있도록 만들어 놓음
  - 주석을 시각화하는 것
  

```python
def double(num):
    """Function to double the value""" # 독파일을 부르면 나오는 값 (설명값)
    return 2*num  # num을 2배 값으로 바꿈
print(double.__doc__)  # 독파일을 출력해서 아래코드에 나옴
print(double(5)) # 5x2=10 값으로 실행
```

```python
help(double) # (.__doc__) 대신 확인가능한 방법
```

```python
# 바로출력
print(1,2,3,4, "t")

# 전체 줄바꿈 출력
print(1,2,3,4,"t", sep="\n")

# 중간만 줄바꿈 출력
print(1,2,3,'\n', 4, "t")

# 끝부분에 원하는 명령을 출력
print(1,2,3, end='\t') # tap을 해줘라는 명령
print(4,5,6)
print(7,8,9, end='%')
```
